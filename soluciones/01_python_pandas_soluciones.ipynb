## 4. Listas, índices y rebanadas

Una **lista** guarda muchos valores en orden, entre corchetes `[ ]`.

- El primer elemento es el **índice 0**.
- Los índices **negativos** cuentan desde el final: `-1` es el último.
- Una **rebanada** `lista[inicio:fin]` incluye `inicio` y **no** incluye `fin`.

````python
notas = [7.8, 9.1, 4.5, 6.0]

print(notas[0])       # el primero
print(notas[-1])      # el último
print(len(notas))     # cuántos hay
print(notas[1:3])     # índices 1 y 2

notas.append(8.3)     # agregar al final
print(notas)
print("Promedio:", sum(notas) / len(notas))

palabra = "HONDURAS"  # el texto también se puede rebanar
print(palabra[0:3], palabra[-3:], palabra[::-1])
````

## 7. Ciclos con `for`

`for n in lista:` significa "para cada `n` dentro de la lista, haz esto". `range(3)` genera `0, 1, 2`.

Un patrón muy común es el **acumulador**: una variable que empieza en 0 y va sumando en cada vuelta.

````python
notas = [8.5, 9.0, 7.5]

total = 0
for n in notas:
    total = total + n
    print("Sumé", n, "→ total =", total)
print("Promedio:", total / len(notas))

for i in range(3):
    print("vuelta", i)
````

## 8. Funciones

Una **función** es una receta reutilizable:

- `def` **define** la receta (todavía no se ejecuta).
- Los **parámetros** son los ingredientes.
- `return` **devuelve** el resultado.

Ya usamos funciones: `print`, `len`, `sum`, `max`, `type`.

````python
def promedio(valores):
    total = sum(valores)
    return total / len(valores)

print(promedio([7.8, 9.1, 4.5]))
print(promedio([10, 8]))
````

### ✏️ Ejercicio 10
Escribe la función `clasificar(nota)` que **devuelva** `"aprobado"` si `nota >= 6` y `"reprobado"` si no.

````python
def clasificar(nota):
    if nota >= 6:
        return "aprobado"
    return "reprobado"
````

### ✏️ Ejercicio 11 · Una función con un ciclo adentro
Escribe `contar_mayores(valores, umbral)` que devuelva **cuántos** valores son **mayores** que `umbral`.

````python
def contar_mayores(valores, umbral):
    cuenta = 0
    for v in valores:
        if v > umbral:
            cuenta += 1
    return cuenta
````

````python
revisar(lambda: contar_mayores([2, 3, 5, 6, 7, 8], 5) == 3 and contar_mayores([], 1) == 0,
        "¡Función correcta!", "Usa un acumulador que empieza en 0 y súmale 1 cuando v > umbral.")
````

## 12. `groupby`: dividir → aplicar → combinar

`df.groupby("grupo")["calificacion"].mean()` hace tres cosas:

1. **Divide** la tabla en un pedazo por cada valor de `grupo`.
2. **Aplica** `mean()` a cada pedazo.
3. **Combina** los resultados en una tabla nueva.

````python
print(df.groupby("grupo")["calificacion"].mean())
print()
df.groupby("grupo").agg(prom_calif=("calificacion", "mean"),
                        max_horas=("horas", "max"),
                        n=("nombre", "count"))
````

### ✏️ Ejercicio 15
1. `asist_por_grupo`: la asistencia **promedio** de cada grupo.
2. `mejor_grupo`: la letra (`"A"` o `"B"`) del grupo con mayor asistencia promedio (pista: `.idxmax()`).
3. `horas_por_resultado`: el promedio de horas de los que aprobaron vs. los que no (agrupa por `"aprobo"`).

````python
asist_por_grupo = df.groupby("grupo")["asistencia"].mean()
mejor_grupo = asist_por_grupo.idxmax()
horas_por_resultado = df.groupby("aprobo")["horas"].mean()
print(asist_por_grupo, mejor_grupo, horas_por_resultado, sep="\n\n")
````

````python
revisar(lambda: abs(asist_por_grupo["A"] - 81.43) < 0.01, "Asistencia promedio del grupo A ≈ 81.4.",
        'df.groupby("grupo")["asistencia"].mean()')
revisar(lambda: mejor_grupo == asist_por_grupo.idxmax(), lambda: f"El grupo con mayor asistencia es {asist_por_grupo.idxmax()}.",
        "Usa .idxmax() sobre la Series anterior.")
revisar(lambda: horas_por_resultado[True] > horas_por_resultado[False], "Los que aprobaron estudian más horas en promedio.",
        'df.groupby("aprobo")["horas"].mean()')
````

## 13. 🏆 Reto final

Responde con código (una o dos líneas cada una):

1. ¿Cuál es el **nombre** del alumno que más horas estudia? → `reto1`
2. ¿Cuántos alumnos tienen calificación entre **5 y 7** (incluidos)? → `reto2` (pista: `.between(5, 7)`)
3. Crea `df_resumen`: una tabla con, por grupo, la calificación **mínima**, **máxima** y la **cantidad de aprobados**.
4. **Bonus:** ¿qué alumno tiene la **mayor diferencia** entre su calificación y el promedio de su grupo? (pista: `df.groupby("grupo")["calificacion"].transform("mean")`).

````python
reto1 = df.loc[df["horas"].idxmax(), "nombre"]
reto2 = df["calificacion"].between(5, 7).sum()
df_resumen = df.groupby("grupo").agg(minima=("calificacion", "min"),
                                     maxima=("calificacion", "max"),
                                     aprobados=("aprobo", "sum"))
diferencia = (df["calificacion"] - df.groupby("grupo")["calificacion"].transform("mean")).abs()
bonus = df.loc[diferencia.idxmax(), "nombre"]
print(reto1, reto2, bonus)
df_resumen
````

````python
revisar(lambda: reto1 == "Kevin", "Kevin estudia 9 horas.", 'Usa df["horas"].idxmax() para encontrar la fila.')
revisar(lambda: reto2 == 6, "6 alumnos entre 5 y 7.", 'df["calificacion"].between(5, 7).sum()')
revisar(lambda: list(df_resumen["aprobados"]) == [4, 3], "Resumen por grupo correcto.", "Usa groupby + agg con min, max y sum.")
````
